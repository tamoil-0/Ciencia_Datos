# **Clustering**

El clustering es una técnica de aprendizaje no supervisado que agrupa datos similares en clusters o grupos.

Para ello, el algoritmo utiliza el Criterio de la Inercia, que indica que el modelo será óptimo, cuando los grupos formados tengan una distancia mínima intra-cluster y, una distancia máxima inter-cluster.

image.png

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA  # Para reducir a 2D


K-MEANS ---> NUMEROS
K-MODES ---> CATEGORICOS
K-PROTOTYPE ---> NUMERICOS CATEGORICOS

# Cargar dataset
iris = load_iris()
X = iris.data  # Características
y_true = iris.target  # Etiquetas reales (solo para comparación, no se usan en no supervisado)

# Reducir a 2D para visualización (usaremos PCA más adelante)

pca = PCA(n_components=2)

X_2d = pca.fit_transform(X)

# Aplicar K-Means con K=3 (sabemos que hay 3 especies en Iris)
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X_2d)
labels = kmeans.labels_
centroids = kmeans.cluster_centers_

# Visualización
plt.figure(figsize=(8, 6))
plt.scatter(X_2d[:, 0], X_2d[:, 1], c=labels, cmap='viridis', s=50)
plt.scatter(centroids[:, 0], centroids[:, 1], c='red', marker='x', s=200, label='Centroides')
plt.title('Clustering K-Means en Dataset Iris (2D)')
plt.xlabel('Componente Principal 1')
plt.ylabel('Componente Principal 2')
plt.legend()
plt.show()
```
